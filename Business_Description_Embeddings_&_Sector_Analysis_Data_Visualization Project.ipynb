#Data Preparation & Cleaning (10%)

Load the dataset with pre-computed embeddings

Handle missing values appropriately (e.g., categorize ETFs and other special securities)

Document all cleaning and preprocessing steps clearly

```python
import pandas as pd

# Load the CSV file (adjust path if needed)
df = pd.read_csv('companyInfo.csv')

# Identify entries with missing Sector & Industry
missing_sector = df['Sector'].isna() & df['Industry'].isna()

# Detect ETFs/funds using keywords in Description
etf_keywords = ['etf', 'fund', 'index', 'investment objective', 'seeks to track', 'trust']
df['is_etf'] = df['Description'].str.lower().apply(
    lambda x: any(keyword in x for keyword in etf_keywords)
)

# Assign 'ETF' to Sector and Industry for detected ETFs
df.loc[missing_sector & df['is_etf'], 'Sector'] = 'ETF'
df.loc[missing_sector & df['is_etf'], 'Industry'] = 'ETF'

# Label remaining unknowns (optional)
df.loc[missing_sector & ~df['is_etf'], 'Sector'] = 'Unknown'
df.loc[missing_sector & ~df['is_etf'], 'Industry'] = 'Unknown'

# Clean whitespace and standardize capitalization
df['Sector'] = df['Sector'].str.strip().str.title()
df['Industry'] = df['Industry'].str.strip().str.title()

# Save the cleaned dataset to a new CSV file
df.to_csv('companyInfo_cleaned.csv', index=False)

# View sector distribution (for checking)
sector_counts = df['Sector'].value_counts()
print(sector_counts)
```

**Key Cleaning Steps:**

1. **Identified Missing Sector/Industry Entries**
   - Filtered rows where both `Sector` and `Industry` were `NaN`.

2. **Detected ETFs Based on Description Keywords**
   - Applied keyword matching (e.g., “etf”, “fund”, “index”, “seeks to track”, “investment objective”) to the `Description` column to infer which entries described ETFs or similar fund structures.

3. **Filled Missing Values**
   - Assigned `'ETF'` to the `Sector` and `Industry` columns for rows identified as funds.
   - For remaining rows with no clear classification, we labeled both fields as `'Unknown'`.

4. **Standardized Formatting**
   - Cleaned extra whitespace in sector/industry names.
   - Applied `.str.title()` to ensure consistent capitalization (e.g., "Financials", not "financials").

5. **Saved the Cleaned Dataset**
   - Exported the cleaned DataFrame as `companyInfo_cleaned.csv` for future use in analysis and visualization.

6. **Sector Distribution Check**
   - Printed value counts of each sector to verify classification coverage and ensure no fields were left missing.


#Explaining Embeddings (15%)

##Required visualization: Create slides showing how 768 dimensions (16x16x3) can be visualized as an RGB icon

Create a clear, visual explanation of what sentence transformer embeddings are

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ast
import re

# Load cleaned CSV
df = pd.read_csv('companyInfo_cleaned.csv')

# Function to fix badly formatted embedding strings
def fix_embedding_string(embedding_str):
    cleaned = re.sub(r'\s+', ' ', embedding_str.strip())  # remove newlines and extra spaces
    cleaned = re.sub(r'(?<=[0-9]) (?=-?[0-9])', ',', cleaned)  # insert commas between numbers
    return cleaned

# Choose one embedding (e.g., the first company)
raw_embedding = df.loc[0, 'embeddings']
fixed_str = fix_embedding_string(raw_embedding)
embedding_vector = np.array(ast.literal_eval(fixed_str))

# Reshape 768-dim vector to 16x16x3
embedding_reshaped = embedding_vector.reshape((16, 16, 3))

# Normalize to RGB range [0, 255] using np.ptp
embedding_rgb = 255 * (embedding_reshaped - embedding_reshaped.min()) / np.ptp(embedding_reshaped)
embedding_rgb = embedding_rgb.astype(np.uint8)

# Plot as RGB icon
plt.figure(figsize=(8, 8))
plt.imshow(embedding_rgb)
plt.axis('off')
plt.title("Visualized Embedding (16x16x3)")
plt.show()
```

##Explain in plain language how embeddings capture semantic meaning from text

Embeddings are a way for machines to understand the meaning of sentences by turning them into numbers. When a model like a sentence transformer processes a business description, it looks at the full sentence and considers the context of each word. Instead of just counting words or looking for keywords, it creates a numerical representation — a vector — that captures the sentence's overall meaning. These embeddings typically have 768 dimensions and act like a unique “semantic fingerprint” for the sentence. What makes them powerful is that similar meanings result in similar vectors. For example, even if two company descriptions use different words — like "makes smartphones" versus "produces mobile electronics" — the model recognizes that they’re talking about the same kind of business and gives them similar embeddings. This allows us to mathematically compare company descriptions and visualize their relationships. In this way, embeddings capture the deep meaning of text, making it possible to group companies by what they do, not just by what words they use. It’s like turning sentences into pictures of meaning — a picture truly worth a thousand words. A sentence embedding is a list of 768 numbers that represent the meaning of a sentence.

In our case, each company’s business description (text) was transformed into an embedding using a model like BERT. These embeddings are used to compare how similar different companies are — not by keywords, but by the full context of what they do.

Think of it as a digital fingerprint of meaning.

##Use the analogy "a picture is worth a thousand words" to show how this dense representation stores meaning

The phrase “a picture is worth a thousand words” perfectly captures how sentence embeddings work. Just like an image can convey rich details — emotions, settings, or relationships — in a single glance, an embedding condenses the meaning of an entire sentence into a dense, compact vector of numbers. In this project, each company’s business description is transformed into a 768-dimensional vector, which we can reshape and visualize as a 16×16×3 RGB image. That image may look abstract, but it’s a visual fingerprint of the company’s purpose and operations. Just as two photos of similar scenes will share visual patterns, two embeddings from similar company descriptions will produce similar visual textures. This shows how meaning is stored in a dense, visualizable form — not by literal words, but by their underlying concepts. With embeddings, we no longer need to read each description word-for-word; the embedding itself is the picture that communicates what the company does.

#Explaining UMAP (15%)

UMAP use a visual like this flow: High-Dimensional Space (768D) -> UMAP Algorithm -> 2D Visualization




What UMAP Does to the Embedding Space

UMAP (Uniform Manifold Approximation and Projection) reduces high-dimensional data — like our 768-dimensional business description embeddings — into a 2D or 3D space.
The goal is to preserve relationships: companies with similar descriptions stay close, while different ones move apart.
This lets us visualize complex patterns (like sector clustering) in a human-readable format.

| Parameter     | Function                                           | Effect on Visualization                          |
| ------------- | -------------------------------------------------- | ------------------------------------------------ |
| `n_neighbors` | Controls local vs global structure                 | Low → shows local clusters; High → global layout |
| `min_dist`    | Controls how tightly points can be packed together | Low → tight clusters; High → spread-out points   |

In our project, we use n_neighbors = 40 and min_dist = 0.01 as a balanced default.
This setup shows both detailed sector clusters and global relationships between industries.

## Show and explain the impact of these key hyperparameters:

`n_neighbors=40`

In business sectors, low n_neighbors might show small, tightly grouped niches like biotech firms inside healthcare, while high n_neighbors merges broader sectors together.

* Controls how much of the local neighborhood is preserved.

* Lower values (e.g. 5–15): UMAP focuses on very local groupings → many tight mini-clusters

* Higher values (e.g. 40–100): UMAP looks at the broader structure → clusters are smoother and more globally connected.

`min_dist=0.01`

With a low min_dist, sector clusters are more compact and easier to separate. A higher min_dist can make everything blend together more, making patterns harder to spot.

* Controls how tightly UMAP packs points together.

* Low min_dist = points can pack very closely → tight, distinct clusters.

* High min_dist = points repel a bit → more spread-out, less distinct clusters.

```python
import numpy as np

# Convert the list of embeddings into NumPy arrays
df0['embedding_array'] = df0['embeddings'].apply(lambda x: np.array(x))

# Stack into a matrix for UMAP input
embedding_matrix = np.vstack(df0['embedding_array'].values)

# Check shape
print(embedding_matrix.shape)  # Should be (num_companies, 768)
```

```python
# Define settings
umap_settings = [
    {'n_neighbors': 10, 'min_dist': 0.01, 'label': 'Local (n=10, dist=0.01)'},
    {'n_neighbors': 40, 'min_dist': 0.01, 'label': 'Balanced (n=40, dist=0.01)'},
    {'n_neighbors': 80, 'min_dist': 0.5, 'label': 'Global (n=80, dist=0.5)'},
]

# Store results
umap_results = []

# Run UMAP for each setting
for setting in umap_settings:
    reducer = umap.UMAP(n_neighbors=setting['n_neighbors'], min_dist=setting['min_dist'], random_state=42)
    embedding_2d = reducer.fit_transform(embedding_matrix)
    umap_results.append((embedding_2d, setting['label']))
```

```python
# Parameter Comparison Side-by-Side
import pandas as pd

fig, axs = plt.subplots(1, 3, figsize=(18, 5))

for i, (coords, label) in enumerate(umap_results):
    axs[i].scatter(
        coords[:, 0],
        coords[:, 1],
        c=pd.Categorical(df0['Sector']).codes,
        cmap='tab20',
        s=10,
        alpha=0.8
    )
    axs[i].set_title(label)
    axs[i].axis('off')

plt.suptitle("UMAP Visualization: Parameter Impact", fontsize=16)
plt.tight_layout()
plt.show()
```

This image compares how UMAP parameters affect the layout of company embeddings.

n=10: Precise clusters, good for spotting outliers

n=40: Balanced and interpretable

n=80: Broad structure, less separation

| **Center** | `n_neighbors = 40`, `min_dist = 0.01` | Balanced between local detail and global structure         |

##Cross-Sector Similarities: Identify industries from different sectors that cluster together and explain why

```python
# If you have 'UMAP_2D_1' and 'UMAP_2D_2', rename them for consistency
df.rename(columns={'UMAP_2D_1': 'UMAP_X', 'UMAP_2D_2': 'UMAP_Y'}, inplace=True)

# Define the groups again just in case
fintech_like = ['AFRM', 'ALLY', 'ADS']
insurance_group = ['AFL', 'AIG', 'AIZ', 'AON', 'ACGL', 'AJG', 'ALL', 'AFG']
etf_funds = ['ACES', 'AGG', 'ADS', 'AEL', 'AGO', 'AMG', 'APAM']

def label_cluster(ticker):
    if ticker in fintech_like:
        return 'Fintech-like Financials'
    elif ticker in insurance_group:
        return 'Insurance Group'
    elif ticker in etf_funds:
        return 'ETF / Investment Fund'
    else:
        return 'Other'

df['CrossSectorGroup'] = df['Ticker'].apply(label_cluster)

# Now plot
import plotly.express as px

fig = px.scatter(
    df,
    x='UMAP_X',
    y='UMAP_Y',
    color='CrossSectorGroup',
    symbol='CrossSectorGroup',
    hover_name='Ticker',
    title='Cross-Sector Similarities in UMAP Space',
    width=900,
    height=650
)
fig.update_traces(marker=dict(size=10))
fig.show()
```
